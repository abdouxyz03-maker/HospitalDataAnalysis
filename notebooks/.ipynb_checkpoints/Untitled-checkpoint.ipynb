```python
import pandas as pd

# Load the dataset using a relative path
# '../data/patients.csv' moves up one folder out of 'notebooks' and into 'data'
df = pd.read_csv("../data/patients.csv")

# Display the first 5 rows of the dataset
df.head()
```

```python
# Print numeric summary (mean, std, min, max, etc.)
print(df.describe())

# Count occurrences for categorical columns
print("\nSex Distribution:")
print(df['Sex'].value_counts())

print("\nDiagnosis Distribution:")
print(df['Diagnosis'].value_counts())
```

```python
import matplotlib.pyplot as plt

# Create a scatter plot of Age vs Cholesterol
plt.figure(figsize=(8, 5))
plt.scatter(df["Age"], df["Cholesterol"], color="blue", alpha=0.7)
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.title("Cholesterol Level vs. Patient Age")
plt.grid(True)
plt.show()
```
